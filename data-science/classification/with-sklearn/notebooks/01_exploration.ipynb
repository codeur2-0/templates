# Churn : exploration et baseline sklearn

Question métier : quels clients prioriser pour une action de rétention ? Nous vérifions le contrat, observons la cible puis entraînons la baseline sans fuite de données.


## 1. Chargement et contrat

On charge les objets du projet : le notebook ne duplique pas la logique métier.


```python
from pathlib import Path

from churn_sklearn.data import CustomerDataLoader
from churn_sklearn.model import ChurnClassifier

frame = CustomerDataLoader(Path("data/raw/customers.csv")).load()
frame.head(), frame["churned"].value_counts(normalize=True)

train = frame.sample(frac=0.75, random_state=42)
test = frame.drop(train.index)
model = ChurnClassifier().fit(train, train["churned"])
{"test_rows": len(test), "probabilities": model.predict_proba(test).round(3).tolist()}
```
