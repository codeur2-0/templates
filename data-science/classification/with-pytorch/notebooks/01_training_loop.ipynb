# Fraude : boucle d entraînement PyTorch

Ce carnet rend visible le cycle tensoriel : validation Pandera, Dataset, DataLoader et diminution de la loss. Les données sont synthétiques et servent uniquement à comprendre les interfaces.


## 1. Chargement et contrat

On charge les objets du projet : le notebook ne duplique pas la logique métier.


```python
from pathlib import Path

from torch.utils.data import DataLoader

from fraud_pytorch.data import TransactionDataLoader, TransactionDataset
from fraud_pytorch.model import FraudMLP, TorchTrainer

frame = TransactionDataLoader(Path("data/raw/transactions.csv")).load()
dataset = TransactionDataset(frame)
model = FraudMLP(hidden_dim=12)
history = TorchTrainer(model, learning_rate=0.01, epochs=5).fit(
    DataLoader(dataset, batch_size=4, shuffle=True)
)
{"rows": len(dataset), "loss_history": [round(value, 4) for value in history]}
```

## Conclusion

La boucle est explicite et remplaçable. Pour une vraie fraude, préférer un split temporel, une loss pondérée et une métrique orientée coût au simple accuracy.
