# Routage de tickets avec spaCy

Nous contrôlons les labels et entraînons le composant textcat sur le petit corpus de tickets. Une prédiction n est jamais envoyée directement à une équipe sans seuil de confiance.


## 1. Chargement et contrat

On charge les objets du projet : le notebook ne duplique pas la logique métier.


```python
from pathlib import Path

from ticket_spacy.data import TicketDataLoader
from ticket_spacy.model import SpacyTicketClassifier

frame = TicketDataLoader(Path("data/raw/tickets.csv")).load()
model = SpacyTicketClassifier(sorted(frame["label"].unique()), dropout=0.2)
losses = model.fit(frame, epochs=3)
model.predict(["My invoice is incorrect"]), losses
```
