## 1. Chargement et contrat

On charge les objets du projet : le notebook ne duplique pas la logique métier.


```python
from pathlib import Path

from sklearn.model_selection import train_test_split

from loan_tensorflow.data import LoanDataLoader
from loan_tensorflow.model import LoanClassifier

frame = LoanDataLoader(Path("data/raw/loans.csv")).load()
train, test = train_test_split(frame, test_size=0.25, random_state=42, stratify=frame["defaulted"])
model = LoanClassifier(hidden_units=16, learning_rate=0.001)
history = model.fit(train, "defaulted", epochs=3, batch_size=4)
{"epochs": len(history.history["loss"]), "test_rows": len(test)}
```
