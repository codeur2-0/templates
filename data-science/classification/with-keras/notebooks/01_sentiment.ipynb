# Sentiment avec Keras

Nous regardons la distribution des trois labels et montrons que la vectorisation fait partie du modèle exporté. Le texte est synthétique et anglophone.


## 1. Chargement et contrat

On charge les objets du projet : le notebook ne duplique pas la logique métier.


```python
from pathlib import Path

from sentiment_keras.data import ReviewDataLoader
from sentiment_keras.model import SentimentClassifier

frame = ReviewDataLoader(Path("data/raw/reviews.csv")).load()
frame["sentiment"].value_counts()
model = SentimentClassifier(vocabulary_size=1000, sequence_length=20, embedding_dim=16)
history = model.fit(frame, "sentiment", epochs=2, batch_size=4)
model.predict(["The support was excellent"]), history.history["loss"]
```

## Conclusion

Un vrai modèle doit séparer les auteurs et les périodes entre train et test pour ne pas surestimer sa généralisation.
