On charge les objets du projet : le notebook ne duplique pas la logique métier.


```python
from pathlib import Path

import pandas as pd

from orders_etl.schemas import OrderSchema
from orders_etl.transform import OrderTransformer

raw = pd.read_csv(Path("data/raw/orders.csv"))
validated = OrderSchema.validate(raw, lazy=True)
clean = OrderTransformer().run(validated)
{
    "raw_rows": len(raw),
    "duplicate_order_ids": int(raw.order_id.duplicated().sum()),
    "clean_rows": len(clean),
    "revenue": clean.net_amount.sum(),
}
```
